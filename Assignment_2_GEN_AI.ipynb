```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
tokenizer=GPT2Tokenizer.from_pretrained("gpt2")
model=GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
prompt = "Artifical intelligance is changing"
```

```python
#Converting into tokens
inputs=tokenizer(prompt,return_tensors="pt")
```

**Baseline**

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True
)
```

**Experiment 1 — Temperature (0.3)**

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.3
)

text = tokenizer.decode(output[0], skip_special_tokens=True)
print(text)
```

**Experiment 2 — Top-k (20)**

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    top_k=20
)

text = tokenizer.decode(output[0], skip_special_tokens=True)
print(text)
```

**Experiment 3 — Top-p (0.60)**

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=0.7,
    top_p=0.60
)

text = tokenizer.decode(output[0], skip_special_tokens=True)
print(text)
```

**Experiment 4 — Repetition Penalty (1.8)**

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=1.5,
    repetition_penalty=1.8
)

text = tokenizer.decode(output[0], skip_special_tokens=True)
print(text)
```

**Experiment 5 — High Temperature (2.0)**

High temperature increases randomness during word selection, so the model is more likely to choose less probable words. This makes the text more creative, but it can also produce illogical or meaningless sentences.

```python
output = model.generate(
    **inputs,
    max_new_tokens=50,
    do_sample=True,
    temperature=2.0
)

text = tokenizer.decode(output[0], skip_special_tokens=True)
print(text)
```

**3 Custom Versions**

Custom Prompt (Story Opening)

A severe monsoon flood devastated a small village in Nepal. As the rain finally stopped, a young volunteer stepped into the muddy streets, determined to help families rebuild their lives.

```python
custom_prompt = "A severe monsoon flood devastated a small village in Nepal. As the rain finally stopped, a young volunteer stepped into the muddy streets, determined to help families rebuild their lives."

inputs = tokenizer(custom_prompt, return_tensors="pt")
```

**Version 1 – Calm & Realistic**

Version 1 (Temperature = 0.5)

The volunteer organized food distribution and helped children return to temporary classrooms. Although many homes were damaged, hope began to grow as neighbors worked together to rebuild their community.

```python
output = model.generate(
    **inputs,
    max_new_tokens=70,
    do_sample=True,
    temperature=0.5
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

**Version 2 – More Creative**

Version 2 (Temperature = 0.8, Top-k = 40)

Carrying medical supplies and warm blankets, the young volunteer inspired others to join the rescue effort. Every repaired bridge and rebuilt house became a symbol of courage and unity across the flood-affected village

```python
output = model.generate(
    **inputs,
    max_new_tokens=70,
    do_sample=True,
    temperature=0.8,
    top_k=40
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

**Version 3 – Diverse Storytelling**

Version 3 (Temperature = 1.0, Top-p = 0.9)

As the sun appeared after days of heavy rain, the village slowly came back to life. People planted new crops, rebuilt schools, and proved that kindness and determination could overcome even the worst natural disaster.

```python
output = model.generate(
    **inputs,
    max_new_tokens=70,
    do_sample=True,
    temperature=1.0,
    top_p=0.9
)

print(tokenizer.decode(output[0], skip_special_tokens=True))
```

**Brief Analysis**

**Changing the generation parameters clearly affected the quality of the text. Lower temperature produced more predictable and coherent sentences, while higher temperature generated more creative but sometimes meaningless content. Top-k limited the vocabulary, making the output more focused. Top-p created balanced and natural sentences, and the repetition penalty reduced repeated words and phrases.**
